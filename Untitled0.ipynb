```python
import sqlite3
import pandas as pd

# Conexión en memoria (se pierde al cerrar el notebook)
conn = sqlite3.connect(":memory:")
cur = conn.cursor() #objeto para ejecutar comandos de SQL

def query(sql, params=()):
    """Ejecuta un SELECT y devuelve un DataFrame para verlo bonito."""
    return pd.read_sql_query(sql, conn, params=params)

def run(sql, params=()):
    """Ejecuta INSERT/UPDATE/DELETE/DDL y hace commit."""
    cur.execute(sql, params)
    conn.commit()
    print("OK ->", sql.strip().splitlines()[0], "...")
```

```python
cur.executescript('''
DROP TABLE IF EXISTS comida;

CREATE TABLE comida (
    id_platillo INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre VARCHAR(50) NOT NULL,
    categoria VARCHAR(30),
    precio DECIMAL(6,2) NOT NULL,
    disponible BOOLEAN DEFAULT TRUE
);
''')
conn.commit()
print("Tabla creada: comida")
```

```python
# INSERT: alta de platillos
run("""INSERT INTO comida (nombre, categoria, precio, disponible) VALUES
    ('Ensalada', 'entrada', 65.00, TRUE),
    ('Sopa de tortilla', 'entrada', 55.00, TRUE),
    ('Milanesa de pollo', 'plato fuerte', 120.00, TRUE),
    ('Tacos al pastor', 'plato fuerte', 95.00, TRUE),
    ('Pastel de chocolate', 'postre', 45.00, TRUE),
    ('Flan napolitano', 'postre', 40.00, FALSE),
    ('Limonada', 'bebida', 30.00, TRUE),
    ('Chokomilk', 'bebida', 25.00, TRUE);""")

query("SELECT * FROM comida")
```

```python
# UPDATE: aumento de precio a la categoría "plato fuerte"
run("UPDATE comida SET precio = precio * 1.10 WHERE categoria = 'plato fuerte';")
query("SELECT id_platillo, nombre, categoria, precio FROM comida WHERE categoria = 'plato fuerte'")
```

```python
# DELETE: eliminación de registros (se retira el platillo no disponible)
run("DELETE FROM comida WHERE disponible = FALSE;")
query("SELECT * FROM comida")
```

```python
cur.executescript('''
DROP TABLE IF EXISTS alumnos;

CREATE TABLE alumnos (
    id_alumno INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre VARCHAR(50) NOT NULL,
    grado VARCHAR(10),
    edad INTEGER NOT NULL,
    promedio DECIMAL(4,2) DEFAULT 0.0
);
''')
conn.commit()
print("Tabla creada: alumnos")
```

```python
# INSERT: alta de alumnos
run("""INSERT INTO alumnos (nombre, grado, edad, promedio) VALUES
    ('Pedro Sánchez', '3ro A', 8, 85),
    ('Ana Torres', '4to B', 9, 89),
    ('Sofía Castillo', '5to B', 10, 95);""")

query("SELECT * FROM alumnos")
```

```python
# UPDATE: aumento de promedio al grado "3ro A"
run("UPDATE alumnos SET promedio = promedio + 5 WHERE grado = '3ro A';")
query("SELECT id_alumno, nombre, grado, promedio FROM alumnos WHERE grado = '3ro A'")
```

```python
# DELETE: eliminación de registros (se retira el alumno de "4to B")
run("DELETE FROM alumnos WHERE grado = '4to B';")
query("SELECT * FROM alumnos")
```

# **EJERCICIO 3 : TIENDITA ESCOLAR**

```python
import sqlite3
import pandas as pd

# Conexión en memoria (se pierde al cerrar el notebook)
conn = sqlite3.connect(":memory:")
cur = conn.cursor() #objeto para ejecutar comandos de SQL

def query(sql, params=()):
    """Ejecuta un SELECT y devuelve un DataFrame para verlo bonito."""
    return pd.read_sql_query(sql, conn, params=params)

def run(sql, params=()):
    """Ejecuta INSERT/UPDATE/DELETE/DDL y hace commit."""
    cur.execute(sql, params)
    conn.commit()
    print("OK ->", sql.strip().splitlines()[0], "...")
```

```python
cur.executescript('''
DROP TABLE IF EXISTS productos;

CREATE TABLE productos (
    id_producto INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre VARCHAR(50) NOT NULL,
    tipo VARCHAR(30),
    precio DECIMAL(6,2) NOT NULL,
    existencia INTEGER DEFAULT 0
);
''')
conn.commit()
print("Tabla creada: productos")
```

```python
# INSERT: alta de productos
run("""INSERT INTO productos (nombre, tipo, precio, existencia) VALUES
    ('Papas fritas', 'snack', 15.00, 50),
    ('Refresco', 'bebida', 18.00, 40),
    ('Chocolate', 'dulce', 12.00, 30),
    ('Cuaderno', 'papelería', 25.00, 20),
    ('Lápiz', 'papelería', 5.00, 100);""")

query("SELECT * FROM productos")
```

```python
# UPDATE: aumento de precio a la categoría "papelería"
run("UPDATE productos SET precio = precio * 1.10 WHERE tipo = 'papelería';")
query("SELECT id_producto, nombre, tipo, precio FROM productos WHERE tipo = 'papelería'")
```

```python
# DELETE: eliminación de registros (se retira el producto con menor existencia)
run("DELETE FROM productos WHERE existencia < 25;")
query("SELECT * FROM productos")
```
